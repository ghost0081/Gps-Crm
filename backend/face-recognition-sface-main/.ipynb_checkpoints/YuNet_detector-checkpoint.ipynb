```python
import numpy as np
import cv2 as cv
import os
```

```python
weights_path = "face_detection_yunet_2023mar.onnx"
score_threshold = 0.5
face_detector = cv.FaceDetectorYN_create(weights_path, "", (0, 0))
face_detector.setScoreThreshold(score_threshold)
```

```python
# Define the path to the folder containing images
folder_path = "images"
output_folder = "output"
# List all files in the folder
files = os.listdir("images")
feat_dict = {}

# Filter out image files (you can add more extensions as needed)
image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tiff']
image_files = [f for f in files if any(f.lower().endswith(ext) for ext in image_extensions)]

for image_file in image_files:
    image_path = os.path.join(folder_path, image_file)
    img = cv.imread(image_path)
    height, width, _ = img.shape
    face_detector.setInputSize((width, height))
    _, faces = face_detector.detect(img)
    faces = faces if faces is not None else []
    #detected_faces_list = detector.detect_faces(img)
    if len(faces) > 0:
        for face1 in faces:
            #box = list(map(int, face1[:4]))
            #x, y, width, height = face1['box']
            x, y, width, height = list(map(int, face1[:4]))
            x2, y2 = x + width, y + height
            cv.rectangle(img, (x, y), (x2, y2), (0, 0, 255), 4)
    # Display the image with detected faces
    cv2.imshow('Detected Faces', img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2

# Load the YuNet model
# model_path = 'yunet.onnx'  # Path to the pre-trained YuNet model (can be in OpenCV models)
# face_detector = cv2.FaceDetectorYN.create(model_path, "", (320, 320))
weights_path = "face_detection_yunet_2023mar.onnx"
score_threshold = 0.5
face_detector = cv.FaceDetectorYN_create(weights_path, "", (0, 0))
face_detector.setScoreThreshold(score_threshold)

# Load the image
image_path = 'images/people.jpg'
img = cv2.imread(image_path)
h, w, _ = img.shape

# Set input size for the face detector
face_detector.setInputSize((w, h))

# Detect faces
_, faces = face_detector.detect(img)

# Check if any faces are detected
if faces is not None:
    for face in faces:
        # Extract face bounding box
        x, y, width, height = face[:4].astype(int)
        cv2.rectangle(img, (x, y), (x + width, y + height), (0, 255, 0), 2)

# Display the image with detected faces
cv2.imshow('Detected Faces', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Load the YuNet model from OpenCV
model = cv2.FaceDetectorYN.create(
    model="face_detection_yunet_2023mar.onnx", # Path to the YuNet model file
    config="",                                # Leave empty unless custom configurations are needed
    input_size=(320, 320),                    # Input size the model expects (you can adjust this)
    score_threshold=0.9,                      # Minimum confidence threshold for detections
    nms_threshold=0.3,                        # Non-maximum suppression threshold
    top_k=5000                                # Max number of faces detected
)

# Open the video file or use 0 for webcam
cap = cv2.VideoCapture("videos/palace.mp4")  # Replace with 0 for webcam

# Check if the video capture is successful
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

# Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Set model's input size to the video frame size
model.setInputSize((width, height))

while True:
    # Read frame from the video
    ret, frame = cap.read()

    # If no more frames, break
    if not ret:
        break

    # Detect faces in the frame
    faces = model.detect(frame)

    if faces[1] is not None:
        for face in faces[1]:
            # Extract the bounding box and landmarks
            x, y, w, h = map(int, face[:4])
            landmarks = face[4:14].reshape(5, 2)

            # Draw the bounding box
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

            # Draw landmarks
            for (lx, ly) in landmarks:
                cv2.circle(frame, (int(lx), int(ly)), 3, (0, 0))
```

```python
import cv2

# Load the YuNet model from OpenCV
model = cv2.FaceDetectorYN.create(
    model="face_detection_yunet_2023mar.onnx", # Path to the YuNet model file
    config="",                                # Leave empty unless custom configurations are needed
    input_size=(320, 320),                    # Input size the model expects (you can adjust this)
    score_threshold=0.9,                      # Minimum confidence threshold for detections
    nms_threshold=0.3,                        # Non-maximum suppression threshold
    top_k=5000                                # Max number of faces detected
)

# Open the video file or use 0 for webcam
cap = cv2.VideoCapture(0)  # Replace with 0 for webcam

# Check if the video capture is successful
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

# Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Set model's input size to the video frame size
model.setInputSize((width, height))

while True:
    # Read frame from the video
    ret, frame = cap.read()

    # If no more frames, break
    if not ret:
        break

    # Detect faces in the frame
    faces = model.detect(frame)

    if faces[1] is not None:
        for face in faces[1]:
            # Extract the bounding box and landmarks
            x, y, w, h = map(int, face[:4])
            landmarks = face[4:14].reshape(5, 2)

            # Draw the bounding box
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

            # Draw landmarks
            for (lx, ly) in landmarks:
                cv2.circle(frame, (int(lx), int(ly)), 3, (0, 0, 255), -1)

    # Show the frame with detections
    cv2.imshow("YuNet Face Detection", frame)

    # Break the loop if the user presses the 'q' key
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```
